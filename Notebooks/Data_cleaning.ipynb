```python
# checking data types and null counts in the dataset
print("DTYPES:")
print(customer_transaction_data.dtypes)
print()
print("NULL COUNTS:")
print(customer_transaction_data.isnull().sum())
```

There are 11,400 rows and 26 columns. Most column data typed as expected (numbers as float64, int64, flags as bool) but amount_src typed as text (object) instead of numeric. Many columns have nulls: timestamp (29), amount_usd (305), fee (295), ip_address (305), ip_country (301), kyc_tier (300), device_trust_score (295)

```python
# amount_src show as text when it should be float. Checking the unique values in the column to see if there are any non-numeric values.

def is_num(x):
    try:
        float(x)
        return True
    except:
        return False

mask = ~customer_transaction_data['amount_src'].apply(is_num)
print("Non-numeric-looking values:", mask.sum())
print(customer_transaction_data.loc[mask, 'amount_src'].unique())
```

There are four values that have comma thousands-separators making pandas unable to parse them as numbers thus making the whole column text instead.

```python
for c in customer_transaction_data.columns:
    print(c, customer_transaction_data[c].nunique())
```

transaction_id should be a unique reference but it has 11,200 unique values across 11,400 rows — fewer than the row count, meaning some transaction IDs are likely diplicated. Also kyc_tier shows 14 unique values despite it being a 3 to 4 category field. channel also shows 12 for what should be 3 entries. These findings indicate messy category labels.

```python
# check duplicate transaction_id values and duplicate rows

dupe_ids = customer_transaction_data['transaction_id'][customer_transaction_data['transaction_id'].duplicated(keep=False)]
print("Rows sharing a duplicated transaction_id:", dupe_ids.shape[0])

print("Fully duplicate rows (every column identical):", customer_transaction_data.duplicated().sum())
```

400 rows share a transaction_id with another row (i.e. 200 transaction_ids appear twice each), and exactly 200 rows are complete duplicates across every column. That means all of these repeats are simple accidental double-entries, not conflicting records - probably safe to drop.

```python
# checking categorical columns for consistency, unique values and their counts

for col in ['kyc_tier', 'channel', 'home_country', 'ip_country']:
    print(f"--- {col} ---")
    print(customer_transaction_data[col].value_counts(dropna=False))
    print()
```

Each of these fields have the categories buried under inconsistent casing, stray whitespace, typos, and placeholder strings - e.g. channel is really just mobile / web / ATM, but appears as 12 raw variants (" mobile", "MOBILE", "mobille", "weeb", etc. ); kyc_tier is really standard / enhanced / low but appears as 14 variants including literal "unknown", "nan", "NAN" strings that are not actually NaN. This needs to be standardised (lowercase, strip whitespace, map typos) before any grouping or modelling can be done.

```python
# checking source_currency and dest_currency for the same kind of issue.

print(customer_transaction_data['source_currency'].value_counts())
print()
print(customer_transaction_data['dest_currency'].value_counts())
```

Both columns seem clean. There are 3 source currencies (USD, GBP, CAD) and 9 destination currencies, no casing/typo issues noticed here. 

```python
# confirming target variable — is_fraud is binary and checking the class balance.

print(customer_transaction_data['is_fraud'].value_counts())
print(customer_transaction_data['is_fraud'].value_counts(normalize=True).reset_index(name='proportion'))
```

is_fraud is binary (0/1), there are no missing values. 10,403 legitimate transactions (91.25%) vs 997 fraud transactions (8.75%) — a meaningfully class imbalance that will require handling later.

```python
# Descriptive statistics of the numeric fields, to spot obviously wrong ranges or negative values where none should exist

num_cols = ['amount_usd', 'fee', 'exchange_rate_src_to_dest', 'account_age_days',
            'device_trust_score', 'chargeback_history_count', 'risk_score_internal',
            'txn_velocity_1h', 'txn_velocity_24h', 'corridor_risk', 'ip_risk_score']
print(customer_transaction_data[num_cols].describe().T)
```

Most fields look okay. However, fee stands out having a minimum of -1 and a maximum of 9999.99 (fees can't sensibly be negative or about £10k on everyday transfers), and ip_risk_score has a maximum of 1.2, which is odd if the score is meant to be between 0 and 1. txn_velocity_1h has a minimum of -1 and you can't have made negative transactions in the last hour.
device_trust_score has a minimum of -0.1 — everything else in that column is between 0 and 1.

```python

cust_counts = customer_transaction_data['customer_id'].value_counts()
print(cust_counts.describe())
print("Customers with exactly 1 transaction:", (cust_counts == 1).sum(), "/ total customers:", len(cust_counts))
print("Largest customer:", cust_counts.index[0], "with", cust_counts.iloc[0], "transactions")
```
