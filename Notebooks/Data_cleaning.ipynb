### Data Dictionary

transaction_id — Unique reference for the transaction.

customer_id — Reference for the customer who made the transaction.

timestamp — Date and time the transaction occurred (UTC).

home_country — Customer's registered/home country.

source_currency — Currency the transaction was sent in.

dest_currency — Currency the transaction was converted to / sent to.

channel — Channel used to initiate the transaction (mobile / web / ATM).

amount_src — Transaction amount in the source currency.

amount_usd — Transaction amount converted to USD.

fee — Fee charged on the transaction.

exchange_rate_src_to_dest — Exchange rate applied between source and destination currency.

device_id — Device used to make the transaction.

new_device — Whether this is the first transaction from this device for this customer.

ip_address — IP address the transaction originated from.

ip_country — Country resolved from the IP address.

location_mismatch — Whether ip_country differs from home_country.

ip_risk_score — Risk score associated with the originating IP address.

kyc_tier — Customer's Know-Your-Customer verification tier.

account_age_days — Age of the customer's account in days at time of transaction.

device_trust_score — Trust score assigned to the device (0–1 scale).

chargeback_history_count — Number of prior chargebacks on the customer's account.

risk_score_internal — Finlora's internal composite risk score for the transaction.

txn_velocity_1h — Number of transactions by the customer in the last hour.

txn_velocity_24h — Number of transactions by the customer in the last 24 hours.

corridor_risk — Risk rating associated with the source-to-destination currency corridor.

is_fraud — Confirmed fraud outcome (1 = fraud, 0 = legitimate).

### DATA CLEANING AND PREPARATION

```python


customer_transaction_data = customer_transaction_data.drop_duplicates().reset_index(drop=True)
```

```python
customer_transaction_data.shape
```

```python
# converting datatype of timestamp to datetime format

customer_transaction_data['timestamp'] = pd.to_datetime(customer_transaction_data['timestamp'], errors='coerce')
# converting datatype of income_usd to float

customer_transaction_data['amount_src'] = pd.to_numeric(customer_transaction_data['amount_src'], errors='coerce')
```

```python
# checking to see data types have been converted successfully.

customer_transaction_data.info()
```

```python
# Rechecking for null values
customer_transaction_data.isnull().sum()
```

```python
# Getting the exchange rate for the currencies in the dataset. 

exchange_rate = customer_transaction_data[customer_transaction_data['amount_usd'].notna()].groupby('source_currency').apply(
    lambda x: (x['amount_usd'] / x['amount_src']).mean()).to_dict()
exchange_rate
```

```python
# Filling the missing values in the amount_usd column using the exchange rate for the source_currency.
customer_transaction_data['amount_usd'] = customer_transaction_data.apply(
    lambda row: row['amount_usd'] if pd.notna(row['amount_usd']) else row['amount_src'] * exchange_rate.get(row['source_currency'], 1),
      axis=1
      )
```

```python
# Standardizing the categorical columns(home_country, ip_country, kyc_tier, channel) to have consistent values. 

# Channel 

customer_transaction_data['channel'] = customer_transaction_data['channel'].str.strip().str.capitalize()

customer_transaction_data['channel'] = customer_transaction_data['channel'].replace({
    'Mobille': 'Mobile',
    'Weeb': 'Web',
})

# Home_country

customer_transaction_data['home_country'] = customer_transaction_data['home_country'].str.strip()

# ip_country

customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].str.strip()
customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].replace(
    {'NAN': np.nan, 'nan': np.nan}
)
customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna('unknown')
print(customer_transaction_data['ip_country'].value_counts(dropna=False))

# kyc tier

customer_transaction_data['kyc_tier'] = customer_transaction_data['kyc_tier'].str.strip().str.lower()
customer_transaction_data['kyc_tier'] = customer_transaction_data['kyc_tier'].replace(
    {'nan': np.nan}
)
customer_transaction_data['kyc_tier'] = customer_transaction_data['kyc_tier'].replace({
    'standrd': 'standard',
    'enhancd': 'enhanced',
})
customer_transaction_data['kyc_tier'] = customer_transaction_data['kyc_tier'].fillna('unknown')
print(customer_transaction_data['kyc_tier'].value_counts(dropna=False))

```

```python
customer_transaction_data['kyc_tier'].value_counts(dropna=False)
```

```python
# filling missing values in the fee column using the median fee for the source_currency.

if 'fee' in customer_transaction_data.columns:
        if 'channel' in customer_transaction_data.columns:
                customer_transaction_data['fee'] = customer_transaction_data.groupby('channel')['fee'].transform(lambda x: x.fillna(x.median()))
                customer_transaction_data['fee'] = customer_transaction_data['fee'].fillna(customer_transaction_data['fee'].median())
```

```python
# filling missing values in the ip_country column with corresponding values from the home_country column

if {'ip_country', 'home_country'}.issubset(customer_transaction_data.columns):
    customer_transaction_data['ip_country'] = customer_transaction_data['ip_country'].fillna(customer_transaction_data['home_country'])
```

```python
# filling missing values in the device_trust_score column using the median device_trust_score per group of new device and kyc_tier if they exist Otherwise, they would  with the overall median.

if 'device_trust_score' in customer_transaction_data.columns:
    if {'new device', 'kyc_tier'}.issubset(customer_transaction_data.columns):
        customer_transaction_data['device_trust_score'] = customer_transaction_data.groupby(['new device', 'kyc_tier'])['device_trust_score'].transform(lambda x: x.fillna(x.median()))
    customer_transaction_data['device_trust_score'] = customer_transaction_data['device_trust_score'].fillna(customer_transaction_data['device_trust_score'].median())
```

```python
# rechecking for null values

customer_transaction_data.isnull().sum()
```

```python
missing_timestamp = customer_transaction_data[customer_transaction_data['timestamp'].isna()]
print("Rows with missing/invalid timestamp:", missing_timestamp.shape[0])
print("is_fraud rate among these rows:", missing_timestamp['is_fraud'].mean())
print("Overall is_fraud rate:", customer_transaction_data['is_fraud'].mean())
```

```python
missing_ip_address = customer_transaction_data[customer_transaction_data['ip_address'].isna()]
print("Rows with missing/invalid IP address:", missing_ip_address.shape[0])
print("is_fraud rate among these rows:", missing_ip_address['is_fraud'].mean())
print("Overall is_fraud rate:", customer_transaction_data['is_fraud'].mean())
```

```python
# dropping ip_address and timestamp columns as they cannot be filled and the percentage of missing values is somewhat low

customer_transaction_data = customer_transaction_data.dropna(subset=['timestamp', 'amount_src', 'ip_address'])

print(customer_transaction_data.shape)
print(customer_transaction_data['is_fraud'].mean())
```
